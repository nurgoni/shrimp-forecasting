```python
import os
import csv

import pandas as pd
```

# Load Data

```python
# check shape of data

data_path = "data/Data"

for data in os.listdir(data_path):
    file = os.path.join(data_path, data)
    
    df = pd.read_csv(file)
    print(f"{data} shape: {df.shape}")
```

```python
# calculate sparse rate

for data in os.listdir(data_path):
    file = os.path.join(data_path, data)
    
    df = pd.read_csv(file)
    sparse_rate = (df.isna().sum().sum() / df.size) * 100
    print(f"{data} sparse rate: {sparse_rate:.2f} %")
```

```python
# see columns

for data in os.listdir(data_path):
    file = os.path.join(data_path, data)
    
    df = pd.read_csv(file)
    print(f"{data} columns: {df.columns}")
```

```python
# remove columns didn't exists in test document

drop_harvests_columns = ["updated_at", "created_at", "id"]
drop_mortalities_columns = ["updated_at", "created_at", "id"]
drop_samplings_columns = ["updated_at", "created_at", "id", "remark"]
drop_cycles_columns = [
    'remark', 
    'created_at', 
    'updated_at', 
    'initial_age', 
    'limit_weight_per_area', 
    'target_cultivation_day',
    'target_size', 
    'extracted_at', 
    'subscription_type', 
    'ordered_at',
    'hatchery_id', 
    'total_seed_type', 
    'hatchery_name', 
    'pond_name'
]
drop_measurements_columns = ['pond_id', 'turbidity']
drop_feed_trays_columns = ["id", "created_at", "updated_at", 'local_feed_logged_at']
```

## Load All Files

```python
os.listdir(data_path)
```

```python
df_cycles = pd.read_csv("data/Data/cycles.csv")
df_fastings = pd.read_csv("data/Data/fastings.csv")
df_fasting = pd.read_csv("data/Data/fasting.csv")
df_feed_trays = pd.read_csv("data/Data/feed_tray.csv")
df_feeds = pd.read_csv("data/Data/feeds.csv")
df_harvest = pd.read_csv("data/Data/harvests.csv")
df_measurements = pd.read_csv("data/Data/measurements.csv")
df_mortalities = pd.read_csv("data/Data/mortalities.csv")
df_samplings = pd.read_csv("data/Data/samplings.csv")
```

```python
df_cycles = df_cycles.rename(columns={"id": "cycle_id"})

df_harvest = df_harvest.drop(columns=drop_harvests_columns)
df_mortalities = df_mortalities.drop(columns=drop_mortalities_columns)
df_samplings = df_samplings.drop(columns=drop_samplings_columns)
df_cycles = df_cycles.drop(columns=drop_cycles_columns)
df_measurements = df_measurements.drop(columns=drop_measurements_columns)
df_feed_trays = df_feed_trays.drop(columns=drop_feed_trays_columns)

df_fasting = pd.concat([df_fasting, df_fastings],axis=0)
df_fasting = df_fasting.drop_duplicates()
```

## combine data

```python
list_df = {
    "df_cycles": df_cycles,
    "df_fasting": df_fasting,
    "df_feed_trays": df_feed_trays,
    "df_feeds": df_feeds,
    "df_harvest": df_harvest,
    "df_measurements": df_measurements,
    "df_mortalities": df_mortalities,
    "df_samplings": df_samplings
}
```

```python
# check dataframe shape

for name, df in list_df.items():
    print(f"{name} shape: {df.shape}")
```

```python
# check unique of cycle id

for name, df in list_df.items():
    print(f"{name} cycle_id unique: {df['cycle_id'].nunique()}")
```

```python
# check if cycle id is only unique on df_cycle

id = set(df_cycles["cycle_id"].unique())

for name, df in list_df.items():
    if name == "df_cycles":
        continue
    else:
        cycle_id = set(df["cycle_id"].unique())
        print(f"{name} cycle_id unique: {cycle_id - id}")
```
